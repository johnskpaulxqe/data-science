## Import Modules 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

```python
import statsmodels.formula.api as smf
```

## Read Data
Read the data from LAST Week

```python
df = pd.read_csv('../week_11/week_11_intro_binary_classification.csv')
```

```python
df.info()
```

## Fit the logistic regression model

```python
fit_glm = smf.logit(formula='y ~ x', data=df).fit()
```

## Predict the training set

Just to be safe though, let's make a COPY of the training set and ADD new columns to the COPY.

```python
df_copy = df.copy()
```

```python
df_copy['pred_probability'] = fit_glm.predict( df )
```

```python
df_copy.info()
```

```python
df_copy
```

```python
df_copy.describe()
```

### Calssify the predictions of the training set
Continue to use the common or default threashold of 50%

```python
df_copy['pred_class'] = np.where(df_copy.pred_probability > 0.5, 1, 0)
```

```python
df_copy.info()
```

```python
df_copy
```

```python
df_copy.nunique()
```

## Accuracy

Accuracy is the PROPORTION of times the model **CORRECTLY** classifies the OBSERVED output!!!

Accuracy is the NUMBER of CORRECT classification DIVIDED by the total number of rows.

```python
df_copy.head()
```

Let's add a new column that stores if the classification is correct

```python
df_copy['correct_class'] = df_copy.y == df_copy.pred_class
```

```python
df_copy
```

```python
df_copy.correct_class.value_counts()
```

We are interested in the PROPORTION of correct classification rather than the COUNT

```python
df_copy.correct_class.value_counts(normalize=True)
```

```python
 df_copy.correct_class.mean()
```

However, I very rarely create a NEW column to store if the classification is correct.

Instead, I calculate the Accuracy by applying the np.mean() function DIRECTLY to the conditional test

```python
np.mean(df_copy.y == df_copy.pred_class)
```

## Summary

The ACCURACY is a PROPOTION between 0 and 1. The closer to 1 Accuracy is the BETTER model performed. The more ACCURATE the model is copared to the observed value.

Accuracy is a SUMMARY STATISTI. It is the NUMBEr of correct classification DIVIDED by the total number of observations!!!
