## Custom summary functions
We can define our own functions and APPLY them to the DataFrame
To demonstrate, let's define our own AVERAGE

```python
def my_avg(x):
    """ assume x is a Pandas series
    and assume x is a numbers"""
    return x.sum()/x.count()
```

```python
df.mean(numeric_only=True)
```

```python
my_avg(df.F)
```

```python
my_avg(df.C)
```

 But to APPLY a custom function to the entire DatFrame...we need to use the .apply() method. So a METHOD is used to APPLY the function to the columns!

```python
df.apply(my_avg)
```

We instead should only apply 'my_avg()' to NUMERIC columns!
A simple way to do that is to SELECT all NUMBERS in the DataFrame.
Pandas has a helper method '.select_dtypes()' that allows you to easily select all columns of a particular data type

```python
df.select_dtypes('number').apply(my_avg)
```

```python
df.mean(numeric_only=True)
```

### Methods for MISSING values
There are specilized and predefined methods dedicated too IDENTIFYING missing entries in a DataFrame

```python
df
```

The .isnull() method converts a seried inot a BOOLEAN! A TRUE value means the entry is MISSING, while a FALSE corresponds to the value being present

```python
df.B.isnull()
```

By applying .isnull() to the ENTIRE DataFrame returns a DATAFRAME of Booleans!

```python
df.isnull()
```

```python
df.isna()
```

Since the result is a DataFrame we can now apply any appropreate summary method to the Boolean DataFrame
such as SUMMING the total number of MISSINGS!

```python
df.isna().sum()
```

If you want the PROPORTION missing apply the .mean() method instead of .sum()

```python
df.isna().mean()
```

 The above is the summary per column...but we can also summarize per row!
 The result shown below is the number of MISSINGS within each ROW!!!

```python
df.isna().sum(axis=1)
```

This is a simple way to identify ROWS with ZERO missing or the ROW is COMPLETE!!!

```python
df.loc[df.isna().sum(axis=1) == 0, :]
```

This idea of removing ALL missings, or remove any row that has at least 1 missing column...is known as creating the **COMPLETE CASES**
This is what happens behind the scenes in a LOT OF MODELING FUNCTIONS!!!

```python
df.shape
```

```python
df.loc[df.isna().sum(axis=1) == 0, :].shape
```

There is a stramlined operation for creating the COMPLETE CASES!

```python
df.dropna()
```
