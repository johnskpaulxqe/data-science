But, we don NOT want to just let these values stay as stray numbers in our notebook.
We want to COLLECT or combine the summary statistics PER GROUP into a new DataFrame!!!
SPLIT-APPLY-COMBINE breaks a dataset based on categories, applies methods to summarize variables within each smaller datasets, and then combines the summary statistics per group into a new easy to use dataframe!

```python
df_E_summary = pd.DataFrame({'E': df.E.unique(),
                             'B_avg':[df_E_aa.B.mean(), df_E_bb.B.mean(), df_E_cc.B.mean(), df_E_dd.B.mean(), df_E_na.B.mean()]})
```

Pandas provides multiple different ways to APPLY summary method to GROUPED dataframes
The '.groupby().aggregate()' apprach is the most flexible yet strightforward way to apply summary method to DIFFERENT COLUMNS!!!

We can pick the method to apply and to pick which column we apply that method to

```python
df.groupby('E', dropna=False).\
           aggregate(B_avg = ('B', 'mean'),
                     B_stc = ('B', 'std'),
                     B_sem = ('B', 'sem'),
                     C_avg = ('C', 'mean'),
                     C_stc = ('C', 'std'),
                     C_sem = ('C', 'sem'),
                     B_nonmissing = ('B', 'size'),
                    B_numrows = ('B', 'count'),
                    F_nunique = ('F', 'nunique'),
                    G_nunique = ('G', 'nunique'),
                    A_nunique = ('A', 'nunique'),).\
reset_index()
```

rather than just diplaying the GROUPED and SUMMARIZED result...let's SAVE or ASSIGN it to a new object

```python
df_E_summary_info = df.groupby('E', dropna=False).\
           aggregate(B_avg = ('B', 'mean'),
                     B_stc = ('B', 'std'),
                     B_sem = ('B', 'sem'),
                     C_avg = ('C', 'mean'),
                     C_stc = ('C', 'std'),
                     C_sem = ('C', 'sem'),
                     B_nonmissing = ('B', 'size'),
                    B_numrows = ('B', 'count'),
                    F_nunique = ('F', 'nunique'),
                    G_nunique = ('G', 'nunique'),
                    A_nunique = ('A', 'nunique'),).\
reset_index()
```

### Realistic example

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.info()
```

```python
titanic.survived.value_counts()
```

```python
titanic.survived.mean()
```

But does the survival rate depnds on the passenger class?

Let's group by 'pclass' and apply methods to summarize the 'survived' column.
This way we can study or explore does the survival rate change across the pclass category?

```python
titanic.groupby('pclass', dropna=False).\
aggregate(num_rows = ('survived', 'size'),
          num_nonmissing = ('survived', 'count'),
          num_survived = ('survived', 'sum'),
          prop_survived = ('survived', 'mean'),
         survive_sem = ('survived', 'sem')).\
reset_index()
```

Let's group by 2 variables just to see what happens.

Let's group by pclass and class

```python
titanic.pclass.value_counts()
```

```python
titanic['class'].value_counts()
```

```python
titanic.groupby(['pclass', 'class'], dropna=False).\
aggregate(num_rows = ('survived', 'size'),
          num_nonmissing = ('survived', 'count'),
          num_survived = ('survived', 'sum'),
          prop_survived = ('survived', 'mean'),
         survive_sem = ('survived', 'sem')).\
reset_index()
```

It can be useful to simply the grouped and summarized result by only focusing on the observed combinations

```python
titanic.groupby(['pclass', 'class'], dropna=False, observed=True).\
aggregate(num_rows = ('survived', 'size'),
          num_nonmissing = ('survived', 'count'),
          num_survived = ('survived', 'sum'),
          prop_survived = ('survived', 'mean'),
         survive_sem = ('survived', 'sem')).\
reset_index()
```
