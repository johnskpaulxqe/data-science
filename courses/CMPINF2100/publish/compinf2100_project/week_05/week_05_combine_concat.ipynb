## Horizontal concatenation
Bindin columns togetehr
The defult axis is 0 - means it will be virtically combine by defalt
To do a horizontal concatenation we go for axis=1

```python
pd.concat([dfa0, dfa1], axis=1)
```

columns repeat
Column names are nolonger unique
Be careful about horizontal combination
Why would we ever horizontally combine?

```python
dfA_left = dfa0.loc[:, dfa0.columns[:3]].copy()
```

```python
dfA_left
```

```python
dfA_right = dfa0.loc[:, dfa0.columns[-2:]].copy()
```

```python
dfA_right
```

```python
 dfA_right.shape
```

The point of horizontally combining is to bring together DIFFERENT columns that have the same number of rows

```python
pd.concat([dfA_left, dfA_right], axis=1)
```

BUT be carefule if you remove the index with horizontal concactenation...you will REMOVE the column names!!!

```python
pd.concat([dfA_left, dfA_right], axis=1, ignore_index=True)
```
