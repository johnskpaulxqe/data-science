### Review of DataFrames
baseball related Dataframe

```python
baseball_dict = {'City':['Pittsburgh', 'Cinicinati', 'Chicago', 'St. Louis', 'Milwaukee'] ,
                'Team':['Pirates', 'Reds', 'Cubs', 'Cardinals', 'Brewers'],
                'Division': 5 * ['Central'],
                'League':5 * ['NL']}
```

```python
baseball_dict
```

```python
 baseball_df = pd.DataFrame(baseball_dict, columns=['League', 'Division', 'City', 'Team'])
```

```python
baseball_df
```

```python
baseball_df['games_back'] = pd.Series([31.5, 27.5, 22.5, 0, 7.5], index=baseball_df.index)
```

```python
baseball_df
```

sort by the games_back column ignore the index and modify  in place

```python
baseball_df.sort_values(['games_back'], ignore_index=True, inplace=True)
```

```python
baseball_df
```

Add two more columns that have values which change down the rows

```python
baseball_df['wins'] = pd.Series([87, 79, 64, 59, 55], index=baseball_df.index)
```

```python
baseball_df['losses'] = pd.Series([63, 70, 85, 90, 94], index=baseball_df.index)
```

```python
baseball_df
```

```python
baseball_df['season'] = 2022
```

```python
baseball_df
```

### Filter Rows
Filtering rows refers to SELECTING rows based on the CONDITIONAL TESTS

```python
baseball_df.loc[ baseball_df.wins > 65, :]
```

```python
baseball_df.loc[ baseball_df.Team == 'Pirates', :]
```

We also saw how to use the OR operator, '|', to find all row whwere the value equals A or B.
Or, the value is ONE of those presented.

```python
baseball_df.loc[ (baseball_df.Team == 'Cardinals') | (baseball_df.Team == 'Brewers'), : ]
```

.isin() method to steramline the | operator

```python
 baseball_df.loc[ baseball_df.Team.isin(['Cardinals', 'Brewers']), :]
```

```python
 baseball_df.loc[ baseball_df.losses.isin([70, 90]), :]
```

```python
top_teams = baseball_df.loc[ baseball_df.games_back < 10, 'Team'].copy().tolist()
```

```python
top_teams
```

```python
baseball_df.loc[ baseball_df.Team.isin( top_teams ), :]
```

## String Pattern Matching

```python
baseball_df.loc[ baseball_df.City == 'Pittsburgh', :]
```
