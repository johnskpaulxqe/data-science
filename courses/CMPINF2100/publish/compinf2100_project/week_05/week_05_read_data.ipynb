## Read or download from a website

TSV file - Tab separated Value

```python
gap_url = 'https://raw.githubusercontent.com/chendaniely/pandas_for_everyone/master/data/gapminder.tsv'
```

```python
type(gap_url)
```

To read in the data we provide the URL web address as a string instad of a file name on our computer
Because we are reading a TAB separated rather than a CSV we need to change the sep (separater) argument

```python
gap_df = pd.read_csv(gap_url, sep='\t')
```

```python
gap_df.shape
```

```python
gap_df.loc[gap_df.country.isin(['United States', 'India'])]
```
