### Begin Exploring Data by Summarizing Pandas Series
We will Explore data before training predictive models. This process is known as Exploratory Data Analysis (EDA)
An important aspect of EDA is knowing how to calculate SUMMARY STATISTICS. This notebook demonstrates how to summarize Pandas Series.

```python
import numpy as np
import pandas as pd
```

## Review NumPy summary methods
Let's create a list of integers and then convert that to a 1D NumPy array.

```python
my_list = [10, 20, 30, 40, 50, 60, 70, 80]
```

Convert to an array

```python
my_array = np.array( my_list)
```

```python
my_array.mean()
```

```python
my_array.var(ddof=1)
```

```python
my_array.std(ddof=1)
```

```python
my_array.min()
```

```python
my_array.max()
```

### Pandas Series - summary methods
Convert the list into a Pandas Series

```python
my_series = pd.Series(my_list)
```

```python
my_series
```

Most of the Pandas Series summary methods work very similarly to their NumPy counterparts!


```python
my_series.mean()
```

```python
my_series.min()
```

```python
my_series.max()
```

Look closely at variance

```python
my_series.var()
```

```python
my_array.var()
```

```python
my_array.var(ddof=1)
```

```python
my_series.std()
```

```python
my_array.std()
```

```python
my_array.std(ddof=1)
```

Pandas CORRECTLY sets 'ddof=1' when the variance or SD are calculated!!!
Pandas calculate the UNBIASED estimates to varance and SD!!!
