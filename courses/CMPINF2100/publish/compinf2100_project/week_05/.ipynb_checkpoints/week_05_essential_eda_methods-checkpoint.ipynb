What are the basic actions that we should perform for any data analysis task?
### Exploratory Data Analysis (EDA)

Know the number of columns and rows

```python
df.shape
```

Know the names of columns

```python
df.columns
```

Know the types of columns

```python
df.dtypes
```

```python
df.info()
```

```python
df.dtypes.value_counts()
```

Know the cells that miss the value

```python
df.isna().sum()
```

```python
df.isna().sum()[df.isna().sum() < 1]
```

```python
df.isna().sum()[df.isna().sum() > 0]
```

Know the number of unique values

```python
df.nunique()
```

.nunique() method does NOT treat MISSINGS as a VALUE, it drops the missing values

```python
df.nunique(dropna=True)
```

By default dropna is True in nunique method and if you turn it to false it will count the missing value

```python
df.nunique(dropna=False)
```

It is always important to begin summarizing the columns

```python
df.mean(numeric_only=True)
```

instead of going for individual mthods form mean, std etc we can have .describe() method

```python
df.describe()
```

by default .describe select only numeric columns we can change that with include

```python
df.describe(include='object')
```

```python
df.describe(include='all') #not recommended
```

The next step is to begin counting the categorical/string columns!

```python
df.A.value_counts()
```

value_counts by default skip the missing values dropna=False will include them 

```python
df.A.value_counts(dropna=False)
```

```python
df.E.value_counts(dropna=False)
```

we can know the proportion of these values using normalize=True

```python
df.E.value_counts(dropna=False, normalize=True)
```

always show the row count before showing proportion

```python
df.H.value_counts(dropna=False)
```

```python
df.H.value_counts(dropna=False, normalize=True)
```

```python
df.F.value_counts()
```

```python
df.F.value_counts(dropna=False, normalize=True)
```
