## Join or Merge
The 'pd.merge()' function JOINS TWO DataFrames.
We need to specify the TWO DataFrames AND we need to specify the column (or columns) that the two are JOINED ON!!!!
The JOINING variable is referred to as the ***KEY***!!!!!

```python
pd.merge(dfA, dfB, on='E')
```

```python
dfA.shape
```

what happened to the missing rows
The answer has to do with the unique values of the KEY between LEFT and RIGHT data sets!!!

```python
dfA
```

The E == 'cc' value is NOT present in the JOINED data set!!!
This is because in addition ot specifying the KEY that we JOIN ON...we also must specify **HOW** the LEFT and RIGHT dataframes are JOINED!!!
By defalt Pandas uses **INNER** joins!!!
The INNER JOIN keeps all values that are consistent for the KEY between the two data sets!!!
or...the KEY VALUES must be shared between the TWO!!!

```python
pd.merge(dfA, dfB, on='E', how='inner')
```

Other types of JOINS
The LEFT JOIN keeps all KEY values present in the LEFT data set

```python
pd.merge(dfA, dfB, on='E', how='left')
```

NaN = Not a number

```python
pd.merge(dfA, dfB, on='E', how='right')
```

For this result RIGHT JOIN and INNER JOIN produce the same results
The LEFT JOIN generated the MISSING vallues
It all depends on the data sets

MISSING values can be "generated" by JOINS!!!
We are bringing together COLUMNS from one data set into another data set
IF we do not have an exact assciation between the columns, the the missing occur.
May to the larger data set as left and smaller as right 
You can aslo use just .merge()

```python
dfA.merge(dfC, on='F', how='inner')
```

we can merge multiple datasetns in one line of code

```python
pd.merge(dfA, dfB, on='E', how='left').merge(dfC, on='F', how='right')
```

The INNER JOIN used by Pandas by default tries to minimize MISSINGS.
Let's assign the results of this JOIN to a new variable

```python
df = pd.merge(dfA, dfB, on='E', how='left').merge(dfC, on='F', how='right')
```
