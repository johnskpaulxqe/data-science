### Review scatter plots - coloring by a third variable

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.info()
```

## Relationships between numeric columns
We know how to create a SCATTER PLOT to visualize the relationship between 2 numeric (continuous) columns in a DataFrame

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm')
plt.show()
```

### Color by a Categorical or Non-Numeric column
We will include a third variable in the scatter plot! This 3rd variable allows us to GROUP BY or CONDITION ON a grouping variable!!!
For e.g. let's color by `species`

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')
plt.show()
```

The default CATEGORICAL or **Qualitative** color pallette is used above. But...we can change the color palette!

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species',
           palette='Dark2')
plt.show()
```

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species',
           palette='Accent')
plt.show()
```

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species',
           palette='Set1')
plt.show()
```

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species',
           palette='colorblind')
plt.show()
```

Most CATEGORICAL or QUALITATIVE color palettes have a LIMITED number of colors to choose from.
The benefit of the default color palette is that it can handle an ENORMOUS categegories

### Color by a NUMERIC column

Sequential color palettes are useful when you are distinguishing LOW to MEDIUM to HIGH values
BUT...you can also distinguish being AWAY from the MID POINT!! That's where DIVERGING color palettes are best!!

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='body_mass_g',
           palette='icefire')
plt.show()
```

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='body_mass_g',
           palette='coolwarm')
plt.show()
```

You can always include categorical variables as FACETS!!! Thus, you can use 4 variables, 3 of which are numeric and 1 being categorical!!

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='body_mass_g',
            col='species',
            palette='coolwarm')
plt.show()
```

A second categorical varialbe can be associated with row facets! This gives 5 variables being displayed within the single scatter plots!!

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='body_mass_g',
            col='species', row='island',
            palette='coolwarm')
plt.show()
```

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='body_mass_g',
            col='species', row='sex',
            palette='coolwarm')
plt.show()
```
