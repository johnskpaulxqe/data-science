### Visualize continuous to continuous relationships via TREND PLOTS

The TREN PLOT is an extension of the scatter plot that adds further "quantitative" information about the relationship between two continuous variables.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

## Read Data

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.info()
```

## Scatter Plot

Fundamental plot to show relationship between two continuous variables

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm')
plt.show()
```

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')
plt.show()
```

## Trend Plots

We can add TREND LINES to visualize the **LINEAR** relationship between the two continuous variables.

TREND LINES correspond to the **BEST FIT LINE** between the two variables.

Later on...we will learn HOW this line is created...but for now let's visualize it and let the line or TREND HELP US interpret the relationship!

The TREND is not created with the 'sns.relplot()' function

Instead, the TREND is created with the FIGURE LEVEL function 'sns.lmplot().

```python
sns.lmplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm')
plt.show()
```

We are viewing an ANTI or NEGATIVE relationship between the two variables!!!

Bu...what happens if we GROUP or CONDITION ON a categorical variable???

Let's COLOR via the hue argument BY species

```python
sns.lmplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')
plt.show()
```

GIVEN or GROUPING BY or CONDITIONING ON the 'species' CHANGES our opinion about the RELATIONSHIP between the 2 continuous variables! GROUPING BY 'species' reveals a POSITIVE relationship between the two!!!

We can of course create the TREND PLOT between 2 different continuous variables!
