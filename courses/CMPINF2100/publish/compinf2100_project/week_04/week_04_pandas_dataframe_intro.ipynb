### convert to DF

### how to sort

```python
baseball_df.sort_values(['Team'])
```

#index retain original '.index' attribute positions)
want to ignore the index

```python
baseball_df.sort_values(['Team'], ignore_index=True )
```

```python
baseball_df.sort_values(['Team'], ignore_index=True, ascending=False)
```

to store it, assign result to new object but force the deep copy

```python
baseball_df_b = baseball_df.sort_values(['Team'], ignore_index=True, ascending=False).copy()
```

```python
baseball_df_b
```

We can also force PANDAS to modify in place using the inplace arguement

```python
baseball_df.sort_values(['Team'], inplace=True)
```
