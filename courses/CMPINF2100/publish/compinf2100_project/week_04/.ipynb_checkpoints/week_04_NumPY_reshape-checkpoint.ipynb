### RESHAPING with NumPy
We need to review DIMENTIONS or SHAPESS in NumPy. What is the difference between row and columns


```python
import numpy as np
```

```python
np.arange(1, 25)
```

```python
x = np.arange(1, 25)
```

```python
x
```

Let's convert or RESHAPE the 1D array into a 2D array!!!
We will literally ADD a dimention to the object and thus modify where a how the vaoues are stored

```python
x
```

```python
x.reshape(12, 2)
```

```python
x.reshape( 12, 2).ndim
```

```python
x.shape
```

```python
x
```

```python
x.reshape(6, 4)
```

```python
x.reshape(6, -1) # if I don't know the rest the keep -1
```

```python
x.reshape(-1, 2)
```

```python
x.reshape(9, -1) # if it is not evenly divided then it will throw error
```
