---

## 🧠 pandas `.loc[]` vs `.iloc[]` — Summary

| Feature                    | `.loc[]`                                         | `.iloc[]`                                  |
| -------------------------- | ------------------------------------------------ | ------------------------------------------ |
| **Purpose**                | Label-based indexing                             | Integer (position)-based indexing          |
| **Syntax**                 | `df.loc[row_labels, column_labels]`              | `df.iloc[row_positions, column_positions]` |
| **Indexing Type**          | Uses **row/column labels** (names, index values) | Uses **integer positions** (0-based)       |
| **Inclusive or Exclusive** | **Inclusive** of both start and end labels       | **Exclusive** of end (like Python slicing) |
| **Examples**               | `df.loc['A']` → row with index `'A'`             | `df.iloc[0]` → first row                   |
| **Slice Example**          | `df.loc['A':'C']` → includes both A and C        | `df.iloc[0:3]` → includes 0,1,2 (not 3)    |
| **Select Multiple**        | `df.loc[['A', 'C'], ['col1', 'col3']]`           | `df.iloc[[0, 2], [1, 3]]`                  |
| **Boolean Masking**        | ✅ Supported                                      | ✅ Supported                                |
| **Assignment**             | ✅ Used to modify data                            | ✅ Used to modify data                      |

---

## 🔍 Examples

Assume:

```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'age': [25, 30, 35, 40],
    'city': ['NY', 'LA', 'Chicago', 'Houston']
}, index=['a', 'b', 'c', 'd'])
```

### 1️⃣ Using `.loc[]` — Label-based

```python
df.loc['b']                     # Row with index label 'b'
df.loc['b', 'city']             # Single value (Bob's city)
df.loc['a':'c', ['name', 'age']]  # Rows a–c, columns name & age
df.loc[df['age'] > 30]          # Filtered by condition
```

### 2️⃣ Using `.iloc[]` — Position-based

```python
df.iloc[1]                      # Second row (Bob)
df.iloc[1, 2]                   # Row 1, column 2 → 'LA'
df.iloc[0:3, 0:2]               # First 3 rows, first 2 columns
df.iloc[[0, 2], [1, 2]]         # Rows 0,2 and columns 1,2
```

---

## ⚡ Key Tips

* `.loc[]` → Think **“by label”** (index/column names)
* `.iloc[]` → Think **“by location”** (numerical positions)
* Both can be used to **get** or **set** data:

  ```python
  df.loc['a', 'age'] = 26
  df.iloc[0, 1] = 26
  ```

```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'age': [25, 30, 35, 40],
    'city': ['NY', 'LA', 'Chicago', 'Houston']
}, index=['a', 'b', 'c', 'd'])
```

The DataFrame looks like this:

| index | name    | age | city    |
| :---- | :------ | --: | :------ |
| a     | Alice   |  25 | NY      |
| b     | Bob     |  30 | LA      |
| c     | Charlie |  35 | Chicago |
| d     | David   |  40 | Houston |

---

### 🔍 Indexing Map (Side-by-side view)

```
     LABEL-BASED (for .loc[])       POSITION-BASED (for .iloc[])
     ─────────────────────────      ────────────────────────────
Row labels:   a   b   c   d          Row positions:  0   1   2   3
Column names: name  age  city        Column positions: 0   1   2
```

---

### 🎯 Examples Compared Side-by-Side

| Operation                 | Using `.loc[]` (by label)          | Using `.iloc[]` (by position)      | Result         |
| ------------------------- | ---------------------------------- | ---------------------------------- | -------------- |
| Select 1st row            | `df.loc['a']`                      | `df.iloc[0]`                       | Alice’s row    |
| Select 2nd column         | `df.loc[:, 'age']`                 | `df.iloc[:, 1]`                    | ages only      |
| Row b, Column city        | `df.loc['b', 'city']`              | `df.iloc[1, 2]`                    | `'LA'`         |
| Rows a–c, cols name & age | `df.loc['a':'c', ['name', 'age']]` | `df.iloc[0:3, 0:2]`                | Subset         |
| Multiple rows (b, d)      | `df.loc[['b', 'd']]`               | `df.iloc[[1, 3]]`                  | Bob, David     |
| Filter condition          | `df.loc[df['age'] > 30]`           | `df.iloc[(df['age'] > 30).values]` | Charlie, David |

---

### 🧠 Quick Rules to Remember

| Concept                             | `.loc[]`                      | `.iloc[]` |
| ----------------------------------- | ----------------------------- | --------- |
| Uses labels (index names)           | ✅                             | ❌         |
| Uses positions (row/column numbers) | ❌                             | ✅         |
| End label in slice included         | ✅ Yes                         | ❌ No      |
| Works with boolean masks            | ✅                             | ✅         |
| Works with numeric range like `0:3` | ❌ (unless labels are numbers) | ✅         |

---

### 🧩 Visual Example in Text

```
DataFrame Grid (row, column):

           name   age   city
index
a (0) →   Alice   25    NY
b (1) →   Bob     30    LA
c (2) →   Charlie 35    Chicago
d (3) →   David   40    Houston

.loc['b', 'city']   → LA   (by label)
.iloc[1, 2]         → LA   (by position)
.loc['a':'c', 'name'] → Alice, Bob, Charlie
.iloc[0:3, 0]         → Alice, Bob, Charlie
```

# Resetting Index
Prefer it to be a regular column if it is storing MEANINGful Values

````python
baseball_df
````

````python
baseball_df.sort_index() # sort menthod
````

````python
baseball_df
````

````python
baseball_df.sort_index( inplace=True)
````

````python
baseball_df
````

We can PULL OUT the .index attribute inot a REGULAR column using the .reset_index() method!

````python
baseball_df.reset_index()
````

````python
baseball_df.reset_index().info()
````

If you don't want to keep the value within the .index attribute when you reset you can DROP them!

````python
baseball_df.reset_index(drop=True)
````

We can rename the column using .rename() method!!! we need to use a DICTIONARY within the columns argument of the .rename() method!!!
The KEY is the original column name and the VALUE is the NEW or desired column name!

````python
baseball_df.reset_index().rename(columns={'index':'games_back'})
````

````python
baseball_df
````

sometimes we want to place each step in a PIPELINE or WORKFLOW of actions on a separate line
This makes the code easier to read especially when there are MANY MANY actions in the WORKFLOW

````python
baseball_df_b = baseball_df.\
reset_index().\
rename(columns={'index':'games_back'})
````

````python
baseball_df_b
````
