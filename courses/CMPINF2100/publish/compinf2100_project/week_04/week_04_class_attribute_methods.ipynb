```python
class myClass:
    """ 
    A simple example for showing how to define custom data types
    """
    # attibute that is integer
    i = 12345
    # attribut string
    s = '12345'
    # attribute list
    a_list = [1,2,3,4,5]
    #methods are defined interally to the class
    #methods are functions linked to a specific data type
    def f(self):
        return 'I need an object to work!'
    # STATIC menthods are different from "regular" methods
    # even though they are defined similar to "regular" methods
    def fstatic():
        return "I work without an object"
```

## Intialization methods
Initialization methods are executed the moment the object is defined or **initialized** Initialization methods will be very important when we get to work with scikit-learn later in the semester
Let us define a NEW user deined data type, Dog. This data type will have hard coded attributes and AND an attribute that the USEr specific

```python
class Dog:
    kind = 'caine'
    feet = 'paws'
    has_tail = 'yes'
    # initialization methods are like functions
    # initialization methods are PRIVATE
    # that means we need DUNDERS!!!!
    def __init__(self, dogs_name):
        self.name = dogs_name
```

Create a new variable or object my_dog that is of the Dog data type

### Define another data type
This time let's use a data type Pet that has more attributes defined in the initialization method

```python
class Pet:
    def __init__(self, species, name, the_pets_age):
        self.species = species
        self.name = name,
        self.age = the_pets_age
```

```python
Pet()
```

```python
my_bros_dog = Pet( species = 'Dog', name = 'Rey', the_pets_age = 4)
```

## Parent and Child classes
We can define custom data type that inherit the methods and attributes of already existing data types
The new data type is refered to as the "child' and the data type we are basing the new one, is referred to as the "parent"
create a new data type a_pet based on Pet

```python
class a_pet(Pet):
    # We only need to define NEW attributes and NEW methods
    # that exist in the CHILD but not the PARENT

    def celebrate_birthday(self):
        self.age +=1
```

```python
geno = a_pet('Dog', 'Geno', 6)
```

```python
geno.age
```

```python
geno.celebrate_birthday()
```

```python
geno.age
```
