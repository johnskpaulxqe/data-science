## Introduction to Visualization
When deciding the type of plot you should use...you must ask yourself the following questions:
* Do I have the data
* What data types am I working with
* What are teh number of unique values for the variables???
* what are my goals???

There are specific plots for different purposes. The typ of plot is MOSTLY determined by the:
* Data type
* Number of unique values
* Individual variables Vs a relationship between multiple variables

## Influence of the data type

There are two main data types we need to concern ourselves with:
* Categorical - strings, letters, words, or non-numeric values
* Continuous - integers, floats, or numeric values
The most basic type of plot for non-numeric variables is the BAR CHART!!!
You should ALWAYS start with creating bar charts!
The bar chart shows the **COUNT** for the number of rows per unique value!!!

The analog for counts with continuous variables or numeric variables are DISTRIBUTIONS. The fundamental type of distribution plot is **HISTOGRAM**.
 
### Influence of the number of unique values

Continuous variables or numric variables with FEW unique values can be EXPLORED as if they are categorical variables!
E.g. the number of cylinders in a car. You will use 4, 6, and 8 as common values for the number of cylinders.

### Individual Vs relationships

A plot focused on an individual variable is commonly called a **MARGINAL** plot. Marginal means we are ignoring every other variable. We are focused on a single variable by itself!

A MARGINAL histogram is a plot of the distribution of a single continuous (numeric) variable.
A MARGINAL bar chart is a plot of the counts of the categories for a single categorical variable
Relationship plots examine how 2 or more variables CHANGE or VARY together!!!
