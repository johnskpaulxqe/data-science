Seanborn works differently than ploting in Pandas.

Seaborn divides functions inot two main categories:

* FIGURE level functions
* AXES level functions

Axes level functions create SPECIFIC kinds of plots.

Figure level functions are 'GOAL ORIENTED'.

Let's begin with AXES leve so we can practice using specific KINDS of plots.

Let's use sns.countplot() function. COUNTING the values and PLOTTING.

Seaborn Axes level functions are built on top of matplotlib and so we need to defind the FIGURE and AXES objects.

The tutor is using an **older version of seaborn**. The behavior changed between versions:

- **Seaborn < 0.12** — bars automatically got different colors by default
- **Seaborn ≥ 0.12** — single color is used unless `hue` is explicitly set

You can check your version with:

```python
import seaborn as sns
print(sns.__version__)
```

If your version is 0.12+, the `hue='sex'` fix I mentioned is the correct modern approach:

```python
sns.catplot(data=penguins, x='sex', kind='count', hue='sex')
plt.show()
```

It produces the same visual result your tutor gets — just requires the extra argument in newer versions.

````python
fig, ax = plt.subplots()

sns.countplot( data = penguins, x = 'sex', ax=ax, hue='sex' )

plt.show()
````

````python
penguins.sex.value_counts(dropna=False)
````

If we want to show a horizontal bar chart in Seaborn we must assign the variable to the y argument!!!

````python
fig, ax = plt.subplots()

sns.countplot(data = penguins, y='sex', ax=ax, hue='sex')

plt.show()
````

But we can create the same figure using the FIGURE LEVEL function instead!!!

We are working with categorical variables and so the gigure level function is 'sns.catplot()'.

The 'sns.catplot() figure level function requires that we specify the kind of plot!

````python
sns.catplot(data = penguins, x='sex', kind='count', hue='sex' )

plt.show()
````

````python
sns.catplot(data = penguins, y='sex', kind='count', hue='sex' )

plt.show()
````

Changing the figure size requires the height and aspect arguments

````python
sns.catplot( data = penguins, y='sex', kind='count',  aspect=2, hue='sex' )

plt.show()
````
