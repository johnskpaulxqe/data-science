#### COUNTING unique values in lists
how to count unique value appearing within in a list

```python
my_list = 7 * ['frodo', 'sam', 'mary', 'pippin']
```

```python
print(len(my_list))
```

```python
my_list
```

### What are Unique values in my_list

```python
unique_list_value = set(my_list)
```

```python
unique_list_value
```

set in python is surrounded by {} but no key/value pair

## How Many entries or record sexist for each unique value in 'my_list'
What are the counts associated with each unique value
the '.count() method will coount the numer of times a specific VALUE appears within a list!

```python
my_list.count('mary')
```

convert set to list

```python
unique_list_value_as_list = list(unique_list_value)
```

```python
type(unique_list_value_as_list)
unique_list_value_as_list
## sort
unique_list_value_as_list.sort()
unique_list_value_as_list
```

Let's use a list comprehension to iterate over the UNIQUE VALUES!!! We will apply the .count() method each UNIQUE and we want to ORGANIZE the results into a new list
`<output> = [<action applied on variable> for <variable> in <sequence>]`

```python
[ my_list.count( unique_value ) for unique_value in unique_list_value_as_list ]
```

but the list comprehension does not provide useful name or context arooun the contents iof the list.
which count correspons to ''sam''?
The DICTIONARY comprehension helps to find it out

```python
{ a_key: my_list.count( a_key ) for a_key in unique_list_value_as_list }
```
