### Why does this matter?
Last week, many of our examples relied on iterating over sequences. Many of thos sequences were based on list! We were collecting values in lists and iterating over them! We **cannot** do the same thing with dictionaries. Dictionaries do not really care about the ORDER the values are entered. Instead, dictionaries care about the KEYS that are ASSOCIATED with the VALUES!

This means we need to think about useful and deceriptive KEY names for the VALUES we are assigning in a dictionary. For example, let's return back to the list of lists. We shold NOT think about just eontaining three lists WITHIN a larger list. Instead we should think about KEY names that DESCRIBE the contents of the lists.

The my_list_of_lists object is printed below. The zeoth list is a list of integers, the first list is a list of str, and the second list is a list of floats.

```python
print( my_list_of_lists )
```

Let's store these three lists a values in a dictionary, my_dict_of_lists . The KEY names are given descriptive names that reflet the cotent of each list. The 'integer_list' contains the list of integers. The 'string_list' contains the list of strings. Lastly, the 'float_list' contains the list of floats.

```python
my_dict_of_lists = { 'integer_list' : [1,2,3],
                    'string_list':['one', 'two', 'three'],
                    'float_list':[1.0, 2.0, 3.0]}
```

The 'my_dict_of_lists' objects has length 3 because it includes 3 KEY/VALUE pairs.

```python
len( my_dict_of_lists )
```

We access the VALUES by providing the KEY names as a string just as with the previous example. The list of integers is accessed and displayed below

```python
my_dict_of_lists['integer_list']
```

Likewise the list of floats is accessed by using its corresponding KEY:

```python
my_dict_of_lists['float_list']
```

The list of strings is accessed by typing its KEY within []

```python
my_dict_of_lists['string_list']
```

I intentionally extracted the VALUES in a different order from how the dictionary was defined. This was to highlight that order does NOT matter with dictionaries! We cannot access the VALUES by using an index because the order position does NOT exist! Thus, the zeroth element and the LAST elements do NOT exist in the dictionary. We cannot identify the "last" VALUE with [-1] like we can in a list because ther is not such thing as FIRST or LAST

```python
my_dict_of_lists[-1]
```

### More details on KEY/VALUE pairs
Example in Section 5.6 of JVG provides an even more detailed overview of working with KEYS and VALUES. I will repeat it now, but with some slight modifications. This example creates a dictionary, monthNumbers, that contains the first 5 months in the calendar year. The KEY/VALUE pairs are created two different ways. The first way uses the month 3-letter abbreviation as the KEY and the month integer number as the VALUE. The second way uses the month integer number as the KEY and the month full name as the VALUE.

monthNumbers = {'Jan':1, 'Feb':2, 'Mar':3, 'Apr':4, 'May':5,
               1:'January', 2:'February', 3:'March', 4:'April', 5:'May'}

The monthNumbers object is a dictionary and has length 10 because 10 KEY/VALUE pairs were typed above.

```python
print( type(monthNumbers) )

print( len(monthNumbers) )
```

The monthNumbers object is a dictionary and so we CANNOT provide an integer index. There is no such thing as the [0] element!

```python
monthNumbers[0]
```

However, we typing monthNumbers[1] does WORK! We do not get an error with this syntax as shown below! Can you see why?

```python
monthNumbers[1]
```

The above line of code does NOT access the first element in monthNumbers. Instead, the above line returns the VALUE associated with the KEY 1. We can access the VALUE associated with another number if it exists as a KEY in the dictionary. For example, the VALUE associated with KEY 3 is returned below.

```python
monthNumbers[3]
```

Returning the VALUE of 1 requires accessing the KEY Jan.

```python
monthNumbers['Jan']
```

Likewise the VALUE of 3 is returned by accessing the KEY Mar.

```python
monthNumbers['Mar']
```

The point of this example is to show the difference between KEY and VALUE. We always return the VALUE associated with the identified KEY!

### Adding ITEMS to a dictionary
An ITEM is a KEY/VALUE pair. We can add more ITEMS by assigning the VALUE to the KEY within the dictionary. Let's add the VALUE 6 to the KEY Jun within monthNumbers. The syntax below looks like we are assigning a value to a "special" variable.

```python
monthNumbers['Jun'] = 6
```

```python
%whos
```

The KEY Jun exists WITHIN the monthNumbers dictionary and its corresponding VALUE is 6!

```python
print(monthNumbers)
```

Look closely at the printed result above. The 'Jun':6 ITEM is located at the end of the dictionary. However, "the end" does not really mean anything to the dictionary. We added a new ITEM to the dictionary and it stores that ITEM. We do NOT access the ITEM using an index like a list. Instead, we access the VALUE by providing the KEY!

```python
monthNumbers['Jun']
```

Let's add one more ITEM with KEY 6 and VALUE 'June'.


```python
monthNumbers[6] = 'June'
```

```python
print( monthNumbers )
```

Again, the new ITEM appears "at the end" but "end" does not mean anything! We access the VALUE through the KEY! The 'June' VALUE is returned because we provided the KEY 6 to monthNumbers. This is NOT the sixth element in the dictionary. After all, the previous print out showed the 6: 'June' ITEM at the END of the dictionary!

```python
monthNumbers[6]
```

### Iterating with dictionaries
Dictionaries are NOT ordered sequences. Thus, dictionary iteration is DIFFERENT from list iteration! Iterating over a dictionary corresponds to iterating over the KEYS and NOT the VALUES! The KEYS are stored in the order they were created within the dictionary. The cell below demnstrates by using using a for-loop to print the KEYS within monthNumbers. The iterating variable is named a_key to explicitly show the sequence is the sequence of KEYS and not the VALUES. The last KEY printed to the screen is the 6 KEY because that the LAST KEY/VALUE pair assigned to the dictionary.

```python
for a_key in monthNumbers:
    print( a_key )
```

However, we can access the VALUE associated with the KEY using []! This works since the iterating variable has the same data type as the KEY! Strings, integers, floats, and other data types can be provided as KEYS within [] to dictionaries. The cell below shows that a_key from the previous for-loop exists in memory as an integer with the value 6.

```python
%whos
```

Using a_key within [] to monthNumbers returns the VALUE associated with the a_key = 6 KEY!

```python
monthNumbers[ a_key ]
```

This means we can programmatically access the VALUES from a dictionary within a for-loop if we so desire.

```python
for a_key in monthNumbers:
    print(monthNumbers[a_key])
```

### check if a key exists
KEYS are obviously VERY important to dictionaries. However, what if we do not know what KEYS are available? For example, what if I do not know if the KEY Oct is present in the dictionary? Using the KEY 'Oct' within [] returns an ERROR because monthNumbers does NOT have any such KEY!

```python
monthNumbers['Oct']
```

It would be very inconvenient if the above was the only way to check if a KEY exists. However, the in operator provides a useful approach to check if a KEY is within a dictionary. The syntax is <KEY> in <dict>. This command returns a boolean. The value True means the KEY IS in the dictionary, while the value False means the KEY is NOT in the dictionary. For example, we can check if 'Oct' is a KEY:

```python
'Oct' in monthNumbers
```

We do NOT raise an error! Instead, the value False is displayed. Checking KEYS that do exist return True, as shown below.

```python
'Jan' in monthNumbers
```

```python
4 in monthNumbers
```

```python
6 in monthNumbers
```

This example uses integers as KEYS to the dictionary. However, I usually do not recommend such a practice. Small integers and floats behave somewhat oddly in Python. For example, the cell above shows 6 is a KEY in the dictionary. The cell below shows that the KEY 6.0 also returns a True!

```python
6.0 in monthNumbers
```

However, 6 and 6.0 are NOT the same data types!

```python
type( 6 ) == type( 6.0 )
```

In summary, the in operator is a "safe" way to check if a KEY exists compared to attempting to access the KEY. As shown below, we will get an error if we access a KEY that does not exist within the dictionary.

```python
monthNumbers['Dec']
```

The in operator does not give an error. Instead, the returned value is False!

```python
'December' in monthNumbers
```

## Sets
Lists and dictionaries are containers. They contain values using different approaches. However, let's now ask a useful data analysis question. Let's assume we have a list of strings, as shown below.

```python
3 * ['a', 'b', 'c']
```

What are the unique values in the above list? This may not seem like a different question. After all, we saw the code that generated the list. The list is also small and so it is easy to COUNT the number of unique values. However, what if we had a bigger list? The list below has 462 elements! Would you want to manually look through such a list to find the unique values?

```python
long_list = 77 * ['a', 'b', 'c', 'd', 'e', 'f']
```

```python
print( len(long_list) )
```

If we did not create long_list, we would NOT be able to easily identify the unique values. The list is simply too long! this is a common and important question in data analysis. Find the number of unique values. Base Python provides a function specially designed to answer this question. The set() function identifies all unique values and returns them in a specialized data type! The set() function is applied to long_list below and the result is assigned to item_set.

```python
item_set = set(long_list)
```

The item_set object is a SET data type!

```python
print( type(item_set) )
```

Printing item_set returns the 6 unique values within curly braces, {}! Thus, a SET "looks" like a dictionary because both SETS and dictionaries use {}. However, a SET is different from a dictionary!

```python
print( item_set )
```

```python
len( item_set )
```

The %whos magic command shows that the SET object has a specific SET of values!

```python
%whos
```

SETS are especially useful in searching operations. We could search through the entire long_list to find if a specific value is contained in that list. For example, let's search for the value 'z'. As shown below, the in operator returns False because 'z' is NOT in long_list. The in operator searched across all 462 elements of long_list to reach this conclusion.

```python
'z' in long_list
```

Instead, we can search through the SET of unique values associated with long_list! This search only required searching over 6 values!

```python
'z' in item_set
```

It does not really matter which approach we use in our current simple example. However, if the original list consisted of millions to tens of millions elements we would need to search over every single element of that very long list. If there were just a few thousand unique values, the search over the SET would be considerably faster!
