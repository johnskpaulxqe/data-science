### Methods for dictionaries
The List Vs Dictionary Jupyter notebook demonstrated several important aspects associated with dictionaries using a modified example from the JVG Book. Let's return to that example here to learn about the methods associated with dictionaries. This example consists of a dictionary, monthNumbers, that contains two styles of KEY/VALUE pair ITEMS. The first style uses the 3 letter month abbrevation as the KEY and the month integer number as the VALUE. The second style uses the month integer number as the KEY and the month full name as the VALUE. The monthNumbers dictionary is defined below.

```python
monthNumbers = {'Jan':1, 'Feb':2, 'Mar':3, 'Apr':4, 'May':5,
               1:'January', 2:'February', 3:'March', 4:'April', 5:'May'}
```

Returning values with .get()
Return the VALUE associated with a KEY by providing the KEY name within []

```python
print( monthNumbers['Jan'] )

print( monthNumbers[1])
```

However, if the KEY does not exist we will get an error. Returning the value for the KEY 'Dec' returns an ERROR because the KEY 'Dec' does NOT exist!

```python
print( monthNumbers['Dec'] )
```

A safer approach is to use the .get() method.

The .get() method will NOT crash and will NOT throw an error if the KEY does NOT exist. By default, the value None is returned. The .get() method therefore tells us no VALUE exists for the KEY 'Dec' because 'Dec' is not a defined KEY!

```python
print( monthNumbers.get('Dec') )
```

Please note that we need to apply the print() function to .get() if we want to display a result if a KEY does NOT exist. For example, nothing is displayed to screen when the .get() method is applied with the 'Dec' KEY below.

```python
monthNumbers.get('Dec') 
```

This is different than what happens if .get() is applied to the KEY 'Jan'. The VALUE 1 is returned because the 'Jan' KEY exists and so its associated VALUE is returned!

```python
monthNumbers.get('Jan')
```

The second argument to the .get() method allows us to specify the value to be returned if the KEY does not exist. This can be useful for error handling purposes.

```python
print( monthNumbers.get('Dec', 'key NOT in the dictionary') )
```

key NOT in the dictionary
The second argument is displayed even without using the print() function.

```python
monthNumbers.get('Dec', 'key NOT in the dictionary')
```

This is because the STRING assigned to the second argument in .get() is returned rather than the None value. The None value is a special data type in Python. As shown below None is NOT a string. The value None is a data type that represents NOTHING!

```python
type( None )
```

### .keys() vs .values()

The .keys() method returns all KEYs in the dictionary.

```python
print( monthNumbers.keys() )
```

The .values() method returns all VALUEs in the dictionary.


```python
print( monthNumbers.values() )
```

The returned object from .keys() is a special data type. It is a **VIEW** type object because it is **LINKED** to the dictionary. This allows us to monitor the state of the object. For example, let's print out the current set of KEYs to the monthNumbers dictionary.

```python
my_keys = monthNumbers.keys()

print( my_keys )
```

The %whos magic command shows my_keys is a dict_keys data type in the Environment. We saw the "value" associated with my_keys in the output above. The %whos output below also shows that the "last" KEY is 5. This is **IMPORTANT** for the next example.

```python
%whos
```

Let's now modify the dictionary by adding the key 'Jun' with value 6

```python
# modify the dictionary
monthNumbers['Jun'] = 6

print(monthNumbers)
```

The dictionary is displayed above. The "last" KEY/VALUE pair is 'Jun': 6. A new KEY is contained in the dictionary. Our previous print out of my_keys did NOT include 'Jun' because the 'Jun' KEY had yet to be added. But what happens if we display my_keys now?

```python
print( my_keys )
```

As we can see above, 'Jun' is the "last" KEY shown! The my_keys object includes the NEW KEY even though this KEY had yet to be added in the dictionary when my_keys was created! This works because my_keys is **NOT** a static object. The my_keys object is a special **VIEW** data type that is **ALWAYS** "watching" or observing another object! Updating that other object changes how "it looks" and thus our VIEW of the object changes!



### Why does this matter?

The .keys() method returns an object that dynamically changes with the dictionary! Changing the dictionary changes ALL VIEW objects associated with that dictionary.

Let's see this idea in action again by DELETING or REMOVING ITEMS using the del operator. The cell below REMOVES the ITEMS associated with the 'Apr', 'Jan', and 1 KEYS.

```python
del monthNumbers['Apr']

del monthNumbers['Jan']

del monthNumbers[1]
```

The my_keys VIEW reflects the changes! Those 3 KEYS are removed!

```python
print( my_keys )
```

The .values() method does NOT include the values associated with the removed KEYS.

```python
monthNumbers.values()
```

Displaying the entire dictionary also shows those 3 KEY/VALUE pairs are removed.

```python
print( monthNumbers )
```

The my_keys object changes dynamically with the dictionary. Iterating over the KEYS within a for-loop means we only iterate over the CURRENTLY defined KEYS! The 3 KEYS previously removed are NOT present in the SEQUENCE iterated over in the for-loop below.

```python
for a_key in my_keys:
    print( monthNumbers[a_key])
```

## Grabbing ITEMS with .items()
A KEY/VALUE pair is called an ITEM. We can extract all ITEMS with the .items() method.

```python
# extract the KEY/VALUES pairs or the ITEMS we use the .items() method
monthNumbers.items()
```

```python
for a_item in monthNumbers.items():
    print( type(a_item) )
```

Tuples CANNOT be modified, so we CANNOT change the values in the ITEMS by applying the .items() method. The .items() method returns tuples so that we can just "see" the ITEMS.
