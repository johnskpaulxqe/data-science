NumPy is dangerous!!!
It's made by a company, used by millions of people in other companies
NumPy uses a wrong method to calculate these
ddof argument must be equal to 1 but it is 0 in NumPy

```python
a[0]
```

```python
a[0].std()
```

```python
a[0].std(ddof=1)
```

NEVER use the DEFAULT `.var()` or `.std()` arguments!!!!
ALWAYS set the ddof=1!!!!

```python
big_array.var(ddof=1)
```

```python
big_array.std(ddof=1)
```

Compare the BIased and unbiased estimators

```python
'BIASED: %1.6f, UNBIASED:%1.6f' % (big_array.var(), big_array.var(ddof=1))
```
