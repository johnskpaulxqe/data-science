### Dictionary comprehensions
Last week, we learned how the **list comprehension** can streamline iteration. We saw how the result of an action is stored within a newly created list by applying that action over a sequence of values. This week, we will continue practicing iteration by learning about the **dictionary comprehension**.

## Review
Let's quickly review the structure of a list comprehension. List comprehensions require identifying the 3 main ingredients of an iteration process. Specifically, those are:

the action to perform
the variable to iterate with
the sequence to iterate over
We unpack those 3 main ingredients and write them "in a single line" using the following structure:

<output> = [ <action to perform using a variable> for <variable> in <sequence> ]

Please note the <> is a placeholder. The word/phrase within <> is what we will change depending on the situation.

The list comprehension allows us to apply nearly any action we want over the elements of a sequence, as long as we want to store the result within a list! For example, let's square all integers from 1 to 11 below.

````python
[ n ** 2 for n in range(1, 12) ]
````

### Storing results in dictionaries
However, what if we do not want to store the result in a list? For example, what if we want to store the result as a dictionary. You might be wondering, why would we want to use a dictionary instead of a list? Dictionaries use names! All values are associated with descriptive names which provides more context to us and anyone else we share our code and results with. It takes more thought and planning to create dictionaries though. We must decide the name or label to assign for the KEY that will be associated with each VALUE. Thus, diciontary comprehensions are more complicated than list comprehensions because a dictionary ITEM consists of TWO things: the KEY and VALUE.

Let's use a simple example to illustrate the steps. The cell below defines a dictionary. This dictionary stores the names of the days of the week as the KEYS. Their associated VALUES are the integers 1 through 5.

````python
ref_dict = {'Monday':1, 'Tuesday':2, 'Wednesday':3, 'Thursday':4, 'Friday': 5}

print(ref_dict)
````

Although this particular example is straightforward to manually type out, imagine if we had 1000s of ITEMS! It would be difficult to type every single ITEM in the dictionary! Instead, let's programmatically create the dictionary from two lists. One list days_of_the_week stores the names we will use as the KEYs. Another list, numbers_of_the_week, stores the VALUEs we will associate with the KEYs.

````python
days_of_the_week = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']

numbers_of_the_week = [1, 2, 3, 4, 5]
````

## Iterating over two sequences
The two lists above contain the same information as ref_dict. However, the information had to be **spread out*. We could not directly associate 'Monday' with 1 because lists DO NOT use names! We need TWO lists because each ITEM in a dictionary has TWO "things": the KEY and the VALUE!

We must therefore iterate over **BOTH** the KEYS and VALUES at the same time. You might be tempted to think that we need to use a NESTED comprehension to do this, but you would be wrong. Let's use a nested for-loop to see why that is the case. The cell below iterates over the elements of days_of_the_week as the outer for-loop. The inner-for loop iterates over the elements in numbers_of_the_week. The two values are combined in a Tuple and printed to the screen. As you can see below, all integer numbers are associated with all day names!

````python
for a_day in days_of_the_week:
    for a_number in numbers_of_the_week:
        print( (a_day, a_number) )
````

We do NOT want the above result. Remember our ref_dict associated 1 only with 'Monday'. The VALUE one was NOT associated with any other day name! We therefore do not want to use a nested iteration structure. Instead we want to iterate over two sequences at the same time!

A simple way to accomplish this is to iterate over the index rather than the values in the list. The index is then used to subset each list accordingly. The cell below demonstrates how to do this by defining the for-loop sequence based on the length of days_of_the_week.

````python
for n in range(len(days_of_the_week)):
    print( (days_of_the_week[n], numbers_of_the_week[n]) )
````

The two lists have the same length so we could have defined the sequence based on numbers_of_the_week.

````python
for n in range(len(numbers_of_the_week)):
    print( (days_of_the_week[n], numbers_of_the_week[n]) )
````

Either way, the result associates one and only one day name with one and only one day number. The integer 1 is ONLY associated with 'Monday'!

Look closely at the either of the two previous cells. We extract the day name and day integer through subsetting. The syntax requires using the square brackets and the iterating variable to pull out the desired values from the lists. The code is rather tedious because the list object names are long. It would be beneficial if we could instead over the elements in BOTH lists at the same time!

That's where the zip() function comes into play. The zip() function "zips up" or combines PAIRS of sequences. We are therefore able to iterate over the values of TWO separate sequences as long as the they have the same length.

Let's see this in action! The cell below creates TWO iterating variables, a_day AND a_value through the zip() function. The two iterating variables are separated by a comma. The two sequences are separated by a comma within the zip() function. The first iterating variable is associated with the first sequence and the second iterating variable is associated with the second sequence. The iterating variables are used within the body of the for-loop to print the day name with the day number!

````python
for a_day, a_value in zip( days_of_the_week, numbers_of_the_week ):
    print( (a_day, a_value) )
````

We can now iterate over two separate sequences! The values within those sequences can be combined as needed in the body of the iteration!

### Creating dictionaries with comprehensions
The dictionary comprehension consists of the same 3 main ingredients as the list comprehension. However, those 3 main ingredients appear TWICE because we must iterate over TWO sequences. Thus, there are 3 main ingredients for the KEYS and 3 main ingredients for the VALUES. The zip() function is essential because we must combine the pair of sequences to iterate over.The other modification is that the comprehension is surrounded by curly braces, {}, because the result is stored in a dictionary.

The dictionary comprehension basic structure is shown below. The action executed within the comprehension is the creation of a dictionary ITEM. The ITEM consists of a KEY and VALUE. The for keyword manages the iteration of the TWO iterating variables. The zip() function creates the PAIR of sequences to iterate over.

<dict> = { <KEY>:<VALUE> for <variable for the KEY>, <variable for the VALUE> in zip(<KEY sequence>, <VALUE sequence>) }

Let's use the dictionary comprehension to create the dictionary of the days of the week. The iterating variables, a_day and a_value are used to define the ITEM's KEY and VALUE, respectively.

````python
my_dict = { a_day: a_value for a_day, a_value in zip(days_of_the_week, numbers_of_the_week) }
````

The my_dict object is a dictionary.

````python
print( type(my_dict) )
````

It has length of 5.

````python
print( len(my_dict) )
````

The day names are KEYS and the day integer numbers of the corresponding VALUES!

````python
print( my_dict )
````

Is that the same result as what we manually defined in ref_dict? The conditional test below shows they are the same!

````python
ref_dict == my_dict
````

We therefore programmatically created the dictionary!

### Why does this matter?
Dictionary comprehensions require careful planning. We are using the same ingredients as the list comprehension, but we need 2 sets of ingredients! Dictionaries provide descriptive KEY names to VALUES. Those names provide useful context to users about the information stored within the dictionary. We will see in the second half of the semester that many predictive modeling functions package the results within dictionaries. We need to be comfortable working with and manipulating dictionaries to be able to work with such results. This example reviewed the dictionary organization by using iteration to programmatically create the dictionary!

### 🔹 **What is a Set in Python**

A **set** is an **unordered collection of unique, hashable elements**.
It is similar to mathematical sets — no duplicates, supports set operations like union, intersection, and difference.

```python
fruits = {"apple", "banana", "cherry"}
```

---

### 🔹 **Mutable Nature**

* Sets are **mutable**, meaning you can **add or remove** elements after creation.
* However, the **elements inside** a set must be **immutable (hashable)** types (e.g., numbers, strings, tuples).

```python
s = {1, 2, 3}
s.add(4)        # modifies the set (mutable)
```

---

### 🔹 **Remove Methods**

| Method       | Description                              | Raises Error if Missing? |
| :----------- | :--------------------------------------- | :----------------------- |
| `remove(x)`  | Removes element `x`                      | ✅ Yes (`KeyError`)       |
| `discard(x)` | Removes element `x`                      | ❌ No                     |
| `pop()`      | Removes and returns an arbitrary element | ✅ Yes (if empty)         |
| `clear()`    | Removes all elements                     | —                        |

```python
s = {1, 2, 3}
s.remove(2)     # removes 2
s.discard(5)    # does nothing if 5 not found
```

---

### 🔹 **Set Operators**

Python supports symbolic operators for common set operations:

| Operator | Meaning                             | Example                            |    |    |
| :------- | :---------------------------------- | :--------------------------------- | -- | -- |
| `        | `                                   | **Union** (elements in either set) | `A | B` |
| `&`      | **Intersection** (elements in both) | `A & B`                            |    |    |
| `-`      | **Difference** (in A but not in B)  | `A - B`                            |    |    |
| `^`      | **Symmetric difference**            | `A ^ B`                            |    |    |
| `<=`     | **Subset**                          | `A <= B` means A ⊆ B               |    |    |
| `>=`     | **Superset**                        | `A >= B` means A ⊇ B               |    |    |

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A | B)  # {1, 2, 3, 4, 5}
print(A - B)  # {1, 2}
print(A <= B) # False (A not subset of B)
```

---

### 🔹 **Hashable Concept**

An object is **hashable** if it has a **hash value that never changes** during its lifetime, and can be compared for equality.
Only hashable objects can be used as elements in a `set` or as keys in a `dict`.

* **Immutable types** → hashable (`int`, `str`, `tuple`)
* **Mutable types** → not hashable (`list`, `dict`, `set`)

```python
hash("apple")    # ✅ works
hash([1, 2, 3])  # ❌ TypeError
```

---

### 🔹 **`__hash__` and `__eq__`**

These are **special methods** that define how objects behave in sets and dictionaries.

* `__hash__()` → returns an integer hash value for the object
* `__eq__(other)` → defines equality comparison

When checking membership (`x in my_set`), Python uses both:

1. `__hash__()` to find the hash bucket
2. `__eq__()` to confirm equality if hash values match

```python
class Person:
    def __init__(self, name):
        self.name = name
    def __hash__(self):
        return hash(self.name)
    def __eq__(self, other):
        return self.name == other.name

p1 = Person("John")
p2 = Person("John")

print(p1 == p2)          # True
print({p1, p2})          # Only one element (unique by hash + eq)
```

### 🔹 Example Demonstration

```python
class Person:
    def __init__(self, name):
        self.name = name
    def __hash__(self):
        print(f"Hashing {self.name}")
        return hash(self.name)
    def __eq__(self, other):
        print(f"Comparing {self.name} with {other.name}")
        return self.name == other.name

p1 = Person("John")
p2 = Person("John")
p3 = Person("Alex")

my_set = {p1, p2, p3}
```

🧩 **Output (simplified):**

```
Hashing John
Hashing John
Comparing John with John
Hashing Alex
```

✅ Result: only **two elements** in the set → `{John, Alex}`
because `p1` and `p2` have the **same hash** and are **equal**.
