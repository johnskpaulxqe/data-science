Creating a 2D array from 1D array requires changing the syntax to np.concatenate()

```python
np.concatenate([[a],[b]])
```

There are speciality functions that simplify STACKING arrays together!!!
Stacking ON top of each other is known as VERTICAL stacking or VERTICAL combing

```python
np.vstack([a,b])
```

Combing array side by side is known as HORIZONTAL stacking

```python
np.hstack([a,b])
```

we can combine or stack more than 2 arrays at the same time
