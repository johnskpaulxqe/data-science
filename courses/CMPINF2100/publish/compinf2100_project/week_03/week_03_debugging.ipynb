## Raising an exception
the %xmode command allows us to diaplay LESS or MORE information around each error

```python
%xmode Plain
```

```python
func1(100,0)
```

what if we used our func1() function within another function?

```python
def func2(x):
    a = x
    b = x - 1
    return func1(a,b)
```

```python
func2(100)
```

```python
func2(10)
```

```python
%xmode verbose
```

```python
func2(1)
```

```python
%xmode Context
```

```python
func2(1)
```
