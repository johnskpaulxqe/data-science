## 🔹 1. Python’s “Names Point to Objects” Model

In Python, **variables are just *names* (or references)** that **point to objects** in memory.
They are *not boxes* that contain values — they are *labels* attached to objects.

Example:

```python
a = [1, 2, 3]
b = a
```

You might think you created two lists, but you didn’t.
You created **one list object** `[1, 2, 3]` in memory, and **two names** (`a` and `b`) that both *point to* it.

Visual model:

```
   a ─────┐
           │
           ▼
        [1, 2, 3]
           ▲
           │
   b ──────┘
```

---

## 🔹 2. Side Effects When Modifying Mutable Objects

If you modify the list using one name, **the change is visible through all names** that point to it.

Example:

```python
a.append(4)
print(b)  # [1, 2, 3, 4]
```

Even though we only called `a.append(4)`, the list changed *in place*,
so `b` also shows the modification — **that’s a side effect**.

---

## 🔹 3. Why It Happens

Python **does not copy objects automatically** when you assign or pass them around.
It just **copies the reference** (i.e., the memory address).

That’s efficient but dangerous if you’re not careful — because now multiple variables refer to the same mutable object.

---

## 🔹 4. Mutability vs. Immutability

* **Mutable objects:** lists, dicts, sets — can be changed *in place*.
  ⇒ prone to side effects.

* **Immutable objects:** ints, floats, strings, tuples — cannot be changed *in place*.
  ⇒ assignments always create new objects.

Example:

```python
x = 10
y = x
x = x + 5
print(y)  # 10, not affected
```

Here, `x` was rebound to a *new integer object* (`15`).
No side effects occur because integers are immutable.

---

## 🔹 5. Side Effects in Function Calls

When you pass a mutable object (like a list) into a function,
and the function modifies it, the caller sees the change.

Example:

```python
def add_item(lst):
    lst.append(99)

nums = [1, 2, 3]
add_item(nums)
print(nums)  # [1, 2, 3, 99]  ← modified!
```

✅ The list `nums` was mutated inside the function.
That’s a **side effect** — the function changed something *outside its own scope*.

---

## 🔹 6. Avoiding Side Effects (Copy Instead)

If you don’t want this to happen, **create a copy** of the list.

```python
a = [1, 2, 3]
b = a[:]      # or list(a) or copy.copy(a)
a.append(4)

print(a)  # [1, 2, 3, 4]
print(b)  # [1, 2, 3]   ← safe copy
```

Now, `a` and `b` refer to *different* objects.

---

## 🔹 7. Deep vs. Shallow Copies

If your list contains other lists (nested structure),
you’ll need a **deep copy** to avoid internal side effects.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)      # shallow copy
c = copy.deepcopy(a)  # deep copy

a[0].append(99)

print(b)  # [[1, 2, 99], [3, 4]]  ← still affected
print(c)  # [[1, 2], [3, 4]]      ← isolated copy
```

---

## 🔹 8. Summary Table

| Concept                    | Example                | Description                              |
| -------------------------- | ---------------------- | ---------------------------------------- |
| **Assignment**             | `b = a`                | Both names point to the same object      |
| **Mutation (side effect)** | `a.append(4)`          | Affects all references                   |
| **Rebinding**              | `a = [5, 6]`           | Creates a new object, doesn’t affect `b` |
| **Shallow copy**           | `b = a[:]`             | New outer list, same inner elements      |
| **Deep copy**              | `b = copy.deepcopy(a)` | Entirely independent copy                |

---

## 🔹 9. Mental Model (How to Think About It)

Think of Python variables as **pointers or sticky notes** attached to real objects:

* `a = [1, 2]` → You stick a note labeled “a” on a list `[1, 2]`.
* `b = a` → You stick another note labeled “b” on the same list.
* Mutating the list changes it for everyone who has a note on it.
