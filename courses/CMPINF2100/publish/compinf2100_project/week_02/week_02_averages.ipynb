### How do we calculate averages for a collection of numbers

```python
sum([1,2,3])
```

```python
len([1,2,3])
```

```python
x_b = [-1, 10, 78, 65, 54 , -678, 1345, 876, 98]
```

```python
sum(x_b)/len(x_b)
```

```python
def my_avg(x):
    """
    this funciton calculates the average
    """
    return sum(x)/len(x)
```

```python
random.seed(2100)
xu = [random.random() for _ in range(11)]
```

```python
my_avg(xu)
```
