### 3.1 Exhaustive Enumeration

````python
#Find the cube root of a perfect cube x = int(input('Enter an integer: ')
x = int(input('Enter an integer: '))
ans = 0
while ans**3 < abs(x):
    # print('Value of the decremnting function abs(x) - ans**3 is', abs(x) - ans**3)
    ans = ans + 1
    print(ans)
if ans**3 != abs(x):
    print(x, 'is not a perfect cube')
else:
    if x < 0:
        ans = -ans
    print('Cube root of', x, 'is', ans)
````

Exhaustive Enumeration means **listing or checking *all possible cases, options, or combinations*** in a systematic way, without skipping any.

It’s often used in mathematics, computer science, and problem-solving when we want to be absolutely certain we’ve considered every possibility.

### Examples:

1. **Mathematics (Combinatorics)**
   To find all possible 3-digit numbers made from {1, 2, 3}, you could exhaustively enumerate:
   123, 132, 213, 231, 312, 321.

2. **Computer Science (Search/Algorithms)**

   * A brute-force password cracker tries every possible password until it finds the correct one — that’s exhaustive enumeration.
   * In graph theory, checking all paths between two nodes to find the shortest one is another example.

3. **Daily Life**
   If you are trying to find a lost key in your house and you check *every single drawer, cupboard, and pocket* without skipping, you’re using exhaustive enumeration.

---

✅ **Key idea**: It guarantees completeness (you don’t miss any solution), but it’s often inefficient for large problems because the number of possibilities can explode.

````python
import math

total = 0

# loop over odd numbers from 3 to 999
for n in range(3, 1000, 2):  
    is_prime = True
    # check divisors up to sqrt(n)
    for divisor in range(2, int(math.sqrt(n)) + 1):
        if n % divisor == 0:
            is_prime = False
            break
    if is_prime:
        total += n

print("Sum of primes greater than 2 and less than 1000:", total)
````

In **Python**, *lambda abstraction* usually refers to the idea of representing a **function anonymously**—without giving it a name. This comes from **lambda calculus**, a mathematical formalism where functions can be treated as first-class objects.

---

### In simple terms:

* A **lambda abstraction** is the definition of a function, usually written as:

  $$
  \lambda \; x . \; f(x)
  $$

  which means *“a function that takes input $x$ and returns $f(x)$”*.

* In Python, we use the `lambda` keyword for this. For example:

  ```python
  square = lambda x: x * x
  ```

  This is a **lambda abstraction**:

  * Input: `x`
  * Output: `x * x`

  It’s equivalent to:

  ```python
  def square(x):
      return x * x
  ```

---

### General form:

```python
lambda arguments: expression
```

* `arguments`: inputs (like function parameters).
* `expression`: a single expression that gets evaluated and returned.

Example:

```python
add = lambda a, b: a + b
print(add(3, 4))  # 7
```

---

### Why is it called abstraction?

In lambda calculus and functional programming:

* *Abstraction* means defining a function without applying it yet.
* When you write `lambda x: x + 1`, you’re **abstracting** over the variable `x`.

So `lambda x: x + 1` represents the **idea** of adding 1, but doesn’t run until you give it an input.

---

✅ **Key takeaway:**
In Python, a *lambda abstraction* is just an **anonymous function** created with the `lambda` keyword, used to define simple functions without a name.

---


````python
def find_root(x, power, epsilon):
    #Find interval containing answer
    if x < 0 and power%2 == 0:
        return None #Negative number has no even-powered roots
    low = min(-1, x)
    high = max(1, x)
    #Use bisection search
    ans = (high + low)/2
    while abs(ans**power - x) >= epsilon:
        if ans**power < x:
            low = ans
        else:
            high = ans
        ans = (high + low)/2
    return ans
````

### 📌 Code again for clarity

```python
def find_root(x, power, epsilon):
    # Find interval containing answer
    if x < 0 and power % 2 == 0:
        return None  # Negative number has no even-powered roots

    low = min(-1, x)
    high = max(1, x)

    # Use bisection search
    ans = (high + low) / 2
    while abs(ans**power - x) >= epsilon:
        if ans**power < x:
            low = ans
        else:
            high = ans
        ans = (high + low) / 2
    return ans
```

---

### 🔎 What this does

It’s a **bisection search algorithm** to find approximate roots of numbers.

1. **Parameters:**

   * `x`: the number we want the root of.
   * `power`: which root (e.g. 2 for square root, 3 for cube root).
   * `epsilon`: tolerance (how close we want the result to be).

2. **Check for impossible cases:**

   ```python
   if x < 0 and power % 2 == 0:
       return None
   ```

   → You can’t take an even root (like square root) of a negative number in real numbers.

3. **Set up search interval:**

   ```python
   low = min(-1, x)
   high = max(1, x)
   ```

   This ensures the true root is always between `low` and `high`.

   * Example: If `x = 25`, then interval is `[1, 25]`.
   * If `x = -27` and `power = 3`, then interval is `[-27, 1]`.

4. **Start bisection:**

   ```python
   ans = (high + low) / 2
   ```

   Take the midpoint of the interval as the guess.

5. **Refine until close enough:**

   ```python
   while abs(ans**power - x) >= epsilon:
   ```

   Keep adjusting the guess until `ans**power` is within `epsilon` of `x`.

6. **Shrink interval each step:**

   * If guess is too small:

     ```python
     if ans**power < x:
         low = ans
     ```
   * If guess is too big:

     ```python
     else:
         high = ans
     ```

7. **Return the approximation:**

   ```python
   return ans
   ```

---

### 🔗 How this relates to **lambda abstraction**

The book uses examples like this to introduce **functions as objects**.

* The **root-finding logic** (`lambda y: y**power - x`) can itself be expressed as a *lambda abstraction*.
* That is, instead of writing a full function, you could describe the problem as finding `y` such that:

  ```python
  f = lambda y: y**power - x
  ```

  And then the task becomes: *find a root of `f(y)`*.

So:

* `find_root` is a named function that uses bisection search.
* But the **idea** behind lambda abstraction is that the *function being solved (`f(y) = y**power - x`)* can be defined anonymously and passed around.

That’s the connection the book tries to make: "functions are just values, and you can abstract them away with `lambda`."

---

✅ **Summary in plain words:**
This function uses **bisection search** to approximate the root of a number `x` for a given power. It keeps shrinking the interval until the guess is close enough (within `epsilon`). The idea of **lambda abstraction** comes in when you realize the function you’re solving (`y**power - x`) doesn’t even need a name — you can represent it anonymously with a `lambda`.

In Python, **scope** determines *where* a variable can be accessed in your program.
The rule Python follows is called **LEGB**, which stands for:

---

### 🔹 LEGB Rule (order of lookup)

When Python sees a variable, it searches in this order:

1. **L – Local**
   Variables defined inside the current function.

   ```python
   def f():
       x = 10  # local
       print(x)
   ```

2. **E – Enclosing**
   Variables in the enclosing function (when you have nested functions).

   ```python
   def outer():
       y = 20  # enclosing
       def inner():
           print(y)  # found in enclosing scope
       inner()
   outer()
   ```

3. **G – Global**
   Variables defined at the top level of the file, or explicitly declared `global`.

   ```python
   z = 30  # global

   def g():
       print(z)  # found in global scope
   g()
   ```

4. **B – Built-in**
   Names preloaded by Python (e.g., `len`, `print`, `range`).

   ```python
   def h():
       print(len([1,2,3]))  # uses built-in len
   h()
   ```

---

### 🔹 Example showing all scopes

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print(x)   # prints "local"
    inner()
    print(x)       # prints "enclosing"

outer()
print(x)           # prints "global"
```

---

### 🔹 Special keywords

* **`global`** – lets you modify a global variable inside a function.
* **`nonlocal`** – lets you modify a variable from an enclosing (but not global) scope.

```python
count = 0

def f():
    global count
    count += 1   # modify global
```

```python
def outer():
    x = 10
    def inner():
        nonlocal x
        x += 5   # modifies enclosing
        print(x)
    inner()
    print(x)
outer()
```

````python
def find_extreme_divisors(n1, n2):
    """Assumes that n1 and n2 are positive ints
    Returns a tuple containing the smallest common divisor > 1 and the largest 
    common divisor of n1 & n2. If no common divisor other thatn 1, returns (NONe, None)"""
    min_val, max_val = None, None
    for i in range(2, min(n1, n2) + 1):
        if n1%i == 0 and n2%i == 0:
            if min_val == None:
                min_val = i
            max_val = i
    return min_val, max_val
min_divisor, max_divisor = find_extreme_divisors(100, 200)    
````

Great question! Let’s break it down clearly:

---

## **1. Lists in Python**

A **list** in Python is a built-in data structure used to store a collection of items in an **ordered** way.

* Lists can contain different data types (integers, strings, floats, even other lists).
* Lists are **indexed** (first element at index 0, second at index 1, and so on).
* Lists are **mutable** (you can change them after creation).

### Example:

```python
# Creating a list
my_list = [10, "hello", 3.14, [1, 2, 3]]

# Accessing items
print(my_list[0])   # 10
print(my_list[1])   # hello
print(my_list[-1])  # [1, 2, 3] (last item)
```

---

## **2. Mutability**

Mutability means whether an object’s contents can be changed after creation.

* **Mutable**: Objects that can be changed in place. (e.g., lists, dictionaries, sets)
* **Immutable**: Objects that cannot be changed once created. (e.g., integers, floats, strings, tuples)

### Example of Mutability in Lists:

```python
nums = [1, 2, 3]

# Modify element
nums[0] = 100
print(nums)  # [100, 2, 3]

# Add elements
nums.append(4)
print(nums)  # [100, 2, 3, 4]

# Remove elements
nums.pop()
print(nums)  # [100, 2, 3]
```

Here, the same list object is being modified instead of creating a new one.

---

## **3. Mutability vs Immutability Example**

```python
# Immutable (string)
s = "hello"
s.upper()
print(s)  # "hello" (unchanged, returns new string)

# Mutable (list)
l = [1, 2, 3]
l.append(4)
print(l)  # [1, 2, 3, 4] (modified in place)
```

---

✅ **Key Takeaway:**

* Lists are **ordered collections** that are **mutable**, meaning you can change their contents (add, remove, or modify items) without creating a new list.
* Mutability is an important concept because it affects how objects behave when passed into functions or assigned to new variables.

## **Aliasing in Python**

**Aliasing** happens when two (or more) variables **refer to the same object in memory**.
Because of this, a change made through one variable will also appear when accessed through the other — since they’re just different “names” (aliases) for the **same object**.

---

### **Example with a List (mutable object):**

```python
list_a = [1, 2, 3]
list_b = list_a   # aliasing: both point to the same list

list_b[0] = 100   # change through list_b
print(list_a)     # [100, 2, 3]  <-- also changed!
print(list_b)     # [100, 2, 3]
```

Here, `list_a` and `list_b` are **aliases** — two names for the same list in memory.

---

### **Aliasing does NOT happen with immutable objects:**

```python
x = 10
y = x    # aliasing attempt
y = 20   # rebinds y to a new int object
print(x) # 10
print(y) # 20
```

Integers are immutable, so when you "change" `y`, Python actually makes `y` refer to a **new object**, leaving `x` unchanged.

---

### **Avoiding Aliasing**

If you want an **independent copy** of a list (not an alias), you need to explicitly copy it:

```python
list_a = [1, 2, 3]
list_b = list_a[:]       # slicing copy
# OR list_b = list_a.copy()
# OR import copy; list_b = copy.deepcopy(list_a)

list_b[0] = 100
print(list_a)  # [1, 2, 3] (unchanged)
print(list_b)  # [100, 2, 3]
```

---

✅ **Key Takeaway:**

* **Aliasing** = multiple variables pointing to the **same mutable object**.
* This can lead to **unexpected side effects** if you modify the object in one place and it changes elsewhere.
* To avoid it, use a **copy** when you need separate objects.

## **Aliasing with Function Arguments**

When you pass a **mutable object** (like a list) into a function, the function parameter becomes an **alias** for the original object.
That means any modification inside the function affects the original list.

---

### Example 1: List Modified Inside a Function

```python
def modify_list(numbers):
    numbers[0] = 999   # modifies the original list
    numbers.append(100)

my_list = [1, 2, 3]
modify_list(my_list)

print(my_list)   # [999, 2, 3, 100]  <-- changed!
```

👉 Inside `modify_list`, `numbers` is just another name (alias) for `my_list`.

---

### Example 2: Reassignment vs Mutation

```python
def reassign_list(numbers):
    numbers = [9, 9, 9]  # new list created, only inside function

def mutate_list(numbers):
    numbers[0] = 9       # mutates the same list

lst = [1, 2, 3]

reassign_list(lst)
print(lst)   # [1, 2, 3] (unchanged)

mutate_list(lst)
print(lst)   # [9, 2, 3] (changed!)
```

👉 Reassignment (`numbers = [9,9,9]`) breaks the aliasing inside the function,
but mutation (`numbers[0] = 9`) affects the original list.

---

### Example 3: Protecting Against Aliasing

If you don’t want the function to modify the original list, pass a **copy**:

```python
def safe_modify(numbers):
    numbers[0] = 0

my_list = [1, 2, 3]
safe_modify(my_list[:])   # pass a copy

print(my_list)  # [1, 2, 3] (unchanged)
```

---

✅ **Key Takeaways**

* Passing a mutable object (like a list) to a function creates an **alias**.
* Mutations inside the function affect the original object.
* Reassignment inside the function does **not** affect the original.
* Use `.copy()` or slicing (`[:]`) if you want to avoid aliasing effects.

## **Python List Methods**

### 🔹 Adding / Inserting

* **`append(x)`** → Adds an item to the end of the list.

  ```python
  lst = [1, 2]
  lst.append(3)   # [1, 2, 3]
  ```

* **`insert(i, x)`** → Inserts an item at index `i`.

  ```python
  lst.insert(1, 99)  # [1, 99, 2, 3]
  ```

* **`extend(iterable)`** → Appends all items from another iterable (list, tuple, etc.).

  ```python
  lst.extend([4, 5])  # [1, 99, 2, 3, 4, 5]
  ```

---

### 🔹 Removing

* **`remove(x)`** → Removes the **first occurrence** of `x`.

  ```python
  lst.remove(99)  # [1, 2, 3, 4, 5]
  ```

* **`pop([i])`** → Removes and returns the item at index `i`. If `i` not given, removes last item.

  ```python
  lst.pop(2)   # removes element at index 2 → returns 3
  ```

* **`clear()`** → Removes all items (empties the list).

  ```python
  lst.clear()   # []
  ```

---

### 🔹 Searching / Counting

* **`index(x[, start[, end]])`** → Returns the first index of `x` (error if not found).

  ```python
  lst = [10, 20, 30, 20]
  lst.index(20)   # 1
  ```

* **`count(x)`** → Returns how many times `x` appears.

  ```python
  lst.count(20)   # 2
  ```

---

### 🔹 Sorting / Reversing

* **`sort(key=None, reverse=False)`** → Sorts the list in place.

  ```python
  nums = [3, 1, 4, 2]
  nums.sort()   # [1, 2, 3, 4]
  nums.sort(reverse=True)  # [4, 3, 2, 1]
  ```

* **`reverse()`** → Reverses the list in place.

  ```python
  nums.reverse()   # [1, 2, 3, 4] → [4, 3, 2, 1]
  ```

---

### 🔹 Copying

* **`copy()`** → Returns a shallow copy of the list.

  ```python
  a = [1, 2, 3]
  b = a.copy()
  ```

### Cloning

---

## **Cloning in Python**

In Python, **cloning** means creating a **copy of an object** so that changes to the new object do not affect the original one.
This is especially important for **mutable objects** (like lists, sets, dictionaries).

---

## **Types of Cloning**

### 1. **Shallow Copy**

* Creates a new object, but does **not** recursively copy nested objects.
* Only the references of inner objects are copied.
* If the original object contains other mutable objects, both copies will still refer to the **same inner objects**.

#### Ways to make a shallow copy:

```python
import copy

# Using slicing
list_a = [1, 2, [3, 4]]
list_b = list_a[:]   # shallow clone

# Using copy() method
list_c = list_a.copy()

# Using copy module
list_d = copy.copy(list_a)

# Modify nested list
list_b[2][0] = 999

print(list_a)  # [1, 2, [999, 4]]
print(list_b)  # [1, 2, [999, 4]]
```

👉 Here, `list_a` and `list_b` share the same inner list `[3,4]`. That’s why both change.

---

### 2. **Deep Copy**

* Creates a new object **and recursively copies** all nested objects.
* Changes in the clone do **not** affect the original, even for nested structures.

#### Example:

```python
import copy

list_a = [1, 2, [3, 4]]
list_b = copy.deepcopy(list_a)

list_b[2][0] = 999

print(list_a)  # [1, 2, [3, 4]]
print(list_b)  # [1, 2, [999, 4]]
```

👉 Here, `list_b` has its own independent copy of `[3,4]`.

---

### 3. **Assignment (NOT cloning!)**

```python
list_a = [1, 2, 3]
list_b = list_a   # alias, not clone

list_b[0] = 100
print(list_a)  # [100, 2, 3]
```

👉 This is just **aliasing** (two names for the same object), not cloning.

---

## ✅ Summary

* **Cloning** = copying an object.
* **Assignment** (`b = a`) → aliasing, not cloning.
* **Shallow Copy** → new object, but shares nested objects. (`a[:]`, `.copy()`, `copy.copy()`)
* **Deep Copy** → full independent copy. (`copy.deepcopy()`)

---


## **1. When Shallow Copy is Enough**

Shallow copy works well when:

* Your data structure is **flat** (no nested mutable objects).
* Or, you don’t care if nested structures are shared between copies.

### Example: Flat List

```python
a = [1, 2, 3, 4]
b = a.copy()    # shallow copy is fine

b[0] = 99
print(a)  # [1, 2, 3, 4] (unchanged)
print(b)  # [99, 2, 3, 4]
```

👉 Since there are no nested lists, shallow copy is enough.

---

## **2. When You Need Deep Copy**

Deep copy is needed when:

* Your object contains **nested mutable structures** (list inside list, dict inside dict, etc.).
* You want to ensure complete independence between the copy and original.

### Example: Nested Dictionary (like JSON data)

```python
import copy

original = {
    "user": "Alice",
    "prefs": {"theme": "dark", "language": "EN"}
}

shallow = original.copy()
deep = copy.deepcopy(original)

# Modify nested dictionary
shallow["prefs"]["theme"] = "light"
deep["prefs"]["language"] = "FR"

print("Original:", original)
# Original: {'user': 'Alice', 'prefs': {'theme': 'light', 'language': 'EN'}}

print("Shallow:", shallow)
# Shallow: {'user': 'Alice', 'prefs': {'theme': 'light', 'language': 'EN'}}

print("Deep:", deep)
# Deep: {'user': 'Alice', 'prefs': {'theme': 'dark', 'language': 'FR'}}
```

👉 The **shallow copy** (`shallow`) still points to the same `"prefs"` dictionary.
👉 The **deep copy** (`deep`) has its own independent `"prefs"`.

---

## **3. Rule of Thumb**

* **Use shallow copy** for:

  * Flat lists, sets, dicts.
  * Performance-sensitive situations (shallow copy is faster).

* **Use deep copy** for:

  * Nested structures (JSON, configs, database records).
  * When you must guarantee independence between original and clone.

---

✅ **Real-World Scenarios**

* **Shallow copy:** copying a flat list of filenames before sorting.
* **Deep copy:** duplicating a nested JSON config so you can safely modify one version without touching the original.

## **What is List Comprehension?**

List comprehension is a **compact way** to create lists in Python using a **single line of code**.
It combines:

* A loop
* An expression (what to put in the list)
* An optional condition

---

## **General Syntax**

```python
[expression for item in iterable if condition]
```

* **expression** → what goes inside the new list
* **item** → variable that takes values from the iterable
* **iterable** → a sequence (list, range, string, etc.)
* **condition (optional)** → filter elements

---

## **Examples**

### 1. Basic Example

```python
nums = [1, 2, 3, 4, 5]
squares = [x**2 for x in nums]

print(squares)  # [1, 4, 9, 16, 25]
```

---

### 2. With Condition

```python
nums = [1, 2, 3, 4, 5, 6]
evens = [x for x in nums if x % 2 == 0]

print(evens)  # [2, 4, 6]
```

---

### 3. Without List Comprehension (long form)

```python
nums = [1, 2, 3, 4, 5]
squares = []
for x in nums:
    squares.append(x**2)
```

👉 List comprehension is shorter and cleaner.

---

### 4. Nested Loops

```python
pairs = [(x, y) for x in [1, 2, 3] for y in [4, 5, 6]]
print(pairs)
# [(1,4), (1,5), (1,6), (2,4), (2,5), (2,6), (3,4), (3,5), (3,6)]
```

---

### 5. String Manipulation

```python
text = "hello"
chars = [ch.upper() for ch in text]
print(chars)  # ['H', 'E', 'L', 'L', 'O']
```

---

### 6. Conditional Expression

```python
nums = [1, 2, 3, 4, 5]
labels = ["even" if x % 2 == 0 else "odd" for x in nums]

print(labels)  # ['odd', 'even', 'odd', 'even', 'odd']
```

---

## ✅ Advantages

* Shorter, cleaner code
* More “Pythonic” (idiomatic)
* Often faster than loops

⚠️ **But:** Overusing nested comprehensions can make code **hard to read**.
