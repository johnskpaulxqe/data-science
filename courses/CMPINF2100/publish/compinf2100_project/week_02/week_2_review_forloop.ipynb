## Nested forloops
- consists of INNER and OUTER forloops

```python
for o1 in range(4):
    print("outer iteration number: %d" %(o1))
    print("---####----")
    for o2 in range(4):
        print("inner iteration number: %d" %(o2))
    print("----$$$$$----")
```
